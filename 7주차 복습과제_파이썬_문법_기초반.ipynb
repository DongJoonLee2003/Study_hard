## while 
- while -> 특정 조건이 만족하는 동안 계속 반복할 때 사용하는 것
- while True 무한루프로 계속 돈다.

- while문은 초기식, 증가에 대한 부분!

```python
# 게임에서 10번 공격
att = 0

while att <7:
    att += 1 
    print(f'나는 키캡 키링을 {att}번 눌렀다.')
    if att ==7:
        print('끝!')
#    0,1,2,3,...,7-> True ->False 7<7
```

```python
# 게임에서 10번 공격
att = 0

while att <10:
    print(f'나는 적을 {att}번 공격했다.')
    if att ==10:
        print('적을 물리쳤다!')
#    0,1,2,3,4,5,...10-> True ->False 10<10
    att += 1 
```

**Plus) 위 코드결과에 'n번째 시행 : ' 이라는 출력문을 덧붙이고 싶을 때는 어떻게 해야할까??**


```python
n = 1

for i in range(4):
    for j in range(3):
        print(f'{n}번째 시행 : {i} 바깥쪽')
        n += 1

        print(f'{n}번째 시행 : {j} 안쪽')
        n += 1
```

```python
## 별을 출력
## *****
## *****
## *****
## *****
## *****
for i in range(5):
    for j in range(5):
        print('*',end='')
    print() # 줄을 바꾸는 것
## 계단식 별을 출력하자
## *
## ** 
## ***
## ****
## *****
print('계단')
for i in range(5):
    for j in range(5):
        if j <= i: 
            print('*',end='')
    print() # 줄을 바꾸는 것

## 역계단
## *****
## ****
## ***
## **
## *
print('역계단')
for i in range(5):
    for j in range(5):
        if j >= i: 
            print('*',end='')
    print() # 줄을 바꾸는 것
    
print('계단')
```

### 간단한 프로그래밍 문제 풀어보기!

### 1에서 100까지 값을 출력!
### 3의 배수는 Fizz
### 5의 배수는 Buzz
### 3과 5의 공배수는 FizzBuzz 출력이 되는 이러한 문제!

```python
for i in range(1,101):
    if i % 15 ==0: #조금 더 상위 개념인 공배수를 먼저 넣는다
        print(f'{i} is FizzBuzz')
    elif i % 3 == 0:
        print(f'{i} is Fizz')
    elif i % 5 == 0:
        print(f'{i} is Buzz')
    else:
        print(f'{i} is Anything')
    
```
