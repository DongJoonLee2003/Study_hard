### 반복문
- loop : 안녕하세요 반복한 것처럼, 규칙적인 똑같은 일을 여러 번 자동으로 해주는 문법
- 왜 필요한가?
    - 같은 코드를 10번, 100번 ,1,000번 복붙 할 수 없다.
    - 데이터들을 차례대로 살펴보고 분석하고 진행하고 싶은데, 데이터가 1억개 -> 1억개를 하나씩 다 쓴다? 불가능하다.

- 반복문의 핵심은 크게 두 가지
    - 몇 번 반복할지 -> for i in range() , for i in 리스트
    - 반복할 때마다 무엇을 할 것인지? 메인코드
    
- for 문 
    - 줄을 세우는 것처럼 차례대로 진행을 한다.
- while 문
    - 조건이 참인 경우 진행

Plus) class_1_score 의 평균 구해보기!

```python
sum = 0
num = 0
for i in class_1_score:
    sum += int(class_1_score[i])
    num += 1
print(sum/num)
```

- 틀린 이유 : i 값이 인덱스의 값이므로 나처럼 class_1_score[i]는 class_1_score[30]이 되므로 리스트 범위를 나가버리게 된다!

```python
sum = 0
num = 0
for i in class_1_score:
    sum += i
    num += 1
print(sum/num)
```

Plus) 구구단 만들자! 1단부터~9단까지!

```python
for i,j in range(1,10):
    print(i,'*',j,'=',i*j)
```

range(1,10) 은 숫자를 하나씩 꺼내준다. (1, 2, 3...)
그런데 for i, j in ... 은 두 개씩 꺼내려고 하니까 충돌이 난 거예요!
구구단은 단 번호(i) 와 곱하는 수(j) 두 개가 필요하니까 → 반복문을 2개 중첩해야 해요!

```python
for i in range(1, 10):     # 단 번호 (1~9)
    print('<',i,'단>')
    for j in range(1, 10):  # 곱하는 수 (1~9)
        print(i, '*', j, '=', i*j)
```

## For , while
- while

- i=0 초기식

- while i < 10: 조건식

- print('hello, world!') 반복할 코드
- i += 1 변화식

```python
j = 0
while j < 3:
    print('초기 j',j)
    print('while문이 돌아가는 중',j)
    j +=1 
    print('변화식 이후 j',j)
```

```python
i = 5
while i > 0:
    print('초기 i',i)
    print('while문이 돌아가는 중',i)
    i -=1 
    print('변화식 이후 i',i)
```

```python
## 주사위 게임

## 주사위는 계속 돌린다. 
## 주사위를 돌려서 -> 3이 나오면 -> 멈춘다!
## 1,2,3,4,5,6

## 난수를 불러오려는 모듈 
import random # random module 사용

random.random()
```

```python
## 주사위의 눈 1,2,3,4,5,6
random.randint(1,6)
```

```python
i = 1
while i != 3: # 주사위가 3이 나온다면 실행을 멈춤 !
    i = random.randint(1,6) # 주사위를 던지는 시행
    print(i)
```
