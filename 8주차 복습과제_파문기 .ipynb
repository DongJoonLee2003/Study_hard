### 3의배수와 5의배수, 15의 배수 -> FizzBuzz 문제!

```python
for i in range(1,10):
    if i % 3 == 0:
        print('Fizz',i)
    elif i % 5 ==0:
        print('Buzz',i)
    else:
        print('Nothing',i)
```

```python
for i in range(1,101):
    if i % 15==0:
        print('FizzBuzz',i)
    elif i % 3 == 0:
        print('Fizz',i)
    elif i % 5 ==0:
        print('Buzz',i)
    else:
        print('nothing',i)
# for i in range(1,101):
#     if i % 5 == 0:
#         print('Buzz',i)
#     elif i % 3 ==0:
#         print('Fizz',i)
```

### 리스트 응용

- 자료구조와 알고리즘
    - 데이터에 대한 형태를 정리한 것
    - 대표적으로 아주 간단한 배열의 데이터를 리스트로 볼 수 있다.
    - 선형적인 자료구조
        - a[0], a[1], a[2], a[3]
            - 인덱스 기준으로 0부터 1씩 증가하는 선형적인 구조
